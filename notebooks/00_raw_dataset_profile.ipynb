## All Numerical Features Detail Analysis

```python
# ============================================================
# NUMERICAL 
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgba

# ------------------------------------------------------------
# 1. Identify numerical features
# ------------------------------------------------------------

numeric_features = df.select_dtypes(include=np.number).columns.tolist()

# Do not analyze the target as a numerical predictor
if TARGET in numeric_features:
    numeric_features.remove(TARGET)

print("=" * 80)
print("NUMERICAL EDA")
print("=" * 80)
print(f"Number of numerical features: {len(numeric_features)}")
print(f"Numerical features: {numeric_features}")


# ------------------------------------------------------------
# 2. Color palette
# ------------------------------------------------------------

FEATURE_COLORS = [
    "#4C78A8",  # blue
    "#F58518",  # orange
    "#54A24B",  # green
    "#E45756",  # red
    "#B279A2",  # purple
    "#FF9DA6",  # pink
    "#72B7B2",  # teal
    "#EDC949",  # yellow
    "#9D755D",  # brown
    "#59A14F",  # green
    "#AF7AA1",  # violet
    "#76B7B2",  # cyan
]

# Additional colors for statistical markers
STAT_COLORS = {
    "mean": "#E63946",
    "median": "#6A1B9A",
    "q1": "#F4A261",
    "q3": "#E76F51",
    "outlier": "#D81B60",
    "grid": "#BDBDBD",
    "text": "#333333",
}


# ------------------------------------------------------------
# 3. Helper functions
# ------------------------------------------------------------

def get_feature_color(index):
    return FEATURE_COLORS[index % len(FEATURE_COLORS)]


def lighten_color(hex_color, amount=0.55):
    """
    Mix a color with white.
    amount = 0   -> original color
    amount = 1   -> white
    """
    rgba = np.array(to_rgba(hex_color))
    rgba[:3] = rgba[:3] + (1 - rgba[:3]) * amount
    return rgba


# ------------------------------------------------------------
# 4. Analyze every numerical feature
# ------------------------------------------------------------

for i, feature in enumerate(numeric_features):

    s = pd.to_numeric(df[feature], errors="coerce").dropna()

    if s.empty:
        print(f"\nSkipping {feature}: no valid numerical observations.")
        continue

    color = get_feature_color(i)
    light_color = lighten_color(color, 0.65)

    # --------------------------------------------------------
    # Basic statistics
    # --------------------------------------------------------

    n = len(s)
    unique = s.nunique()

    minimum = s.min()
    maximum = s.max()
    data_range = maximum - minimum

    mean = s.mean()
    median = s.median()
    mode_values = s.mode()

    if len(mode_values) > 0:
        mode = mode_values.iloc[0]
        mode_count = int((s == mode).sum())
        mode_percentage = (mode_count / n) * 100
    else:
        mode = np.nan
        mode_count = 0
        mode_percentage = 0

    q1 = s.quantile(0.25)
    q3 = s.quantile(0.75)
    iqr = q3 - q1

    std = s.std()
    variance = s.var()

    lower_fence = q1 - 1.5 * iqr
    upper_fence = q3 + 1.5 * iqr

    outliers = s[(s < lower_fence) | (s > upper_fence)]

    zero_count = int((s == 0).sum())
    negative_count = int((s < 0).sum())

    skewness = s.skew()
    kurtosis = s.kurtosis()

    # --------------------------------------------------------
    # Header
    # --------------------------------------------------------

    print("\n\n")
    print("=" * 100)
    print(f"FEATURE {i + 1}: {feature}")
    print("=" * 100)

    # --------------------------------------------------------
    # Basic information
    # --------------------------------------------------------

    print("\n[ BASIC INFORMATION ]")
    print(f"Observations        : {n:,}")
    print(f"Unique values       : {unique:,}")
    print(f"Missing values      : {df[feature].isna().sum():,}")

    # --------------------------------------------------------
    # Range information
    # --------------------------------------------------------

    print("\n[ RANGE INFORMATION ]")
    print(f"Minimum             : {minimum:.4f}")
    print(f"Maximum             : {maximum:.4f}")
    print(f"Range               : {data_range:.4f}")

    # --------------------------------------------------------
    # Central tendency
    # --------------------------------------------------------

    print("\n[ CENTRAL TENDENCY ]")
    print(f"Mean                : {mean:.4f}")
    print(f"Median              : {median:.4f}")
    print(f"Mode                : {mode:.4f}")
    print(f"Mode frequency      : {mode_count:,}")
    print(f"Mode percentage     : {mode_percentage:.2f}%")

    # --------------------------------------------------------
    # Dispersion
    # --------------------------------------------------------

    print("\n[ DISPERSION ]")
    print(f"Standard deviation  : {std:.4f}")
    print(f"Variance            : {variance:.4f}")
    print(f"Q1 (25%)            : {q1:.4f}")
    print(f"Q3 (75%)            : {q3:.4f}")
    print(f"IQR                 : {iqr:.4f}")

    # --------------------------------------------------------
    # Outliers
    # --------------------------------------------------------

    print("\n[ OUTLIER ANALYSIS — IQR METHOD ]")
    print(f"Lower fence         : {lower_fence:.4f}")
    print(f"Upper fence         : {upper_fence:.4f}")
    print(f"Outlier count       : {len(outliers):,}")
    print(f"Outlier percentage  : {(len(outliers) / n) * 100:.2f}%")

    if len(outliers) > 0:
        print("\nOutlier value frequencies:")
        print(
            outliers
            .value_counts()
            .sort_index()
            .to_string()
        )
    else:
        print("No IQR-based outliers detected.")

    # --------------------------------------------------------
    # Zero / negative values
    # --------------------------------------------------------

    print("\n[ ZERO / NEGATIVE VALUES ]")
    print(f"Zero values         : {zero_count:,} "
          f"({zero_count / n * 100:.2f}%)")
    print(f"Negative values     : {negative_count:,} "
          f"({negative_count / n * 100:.2f}%)")

    # --------------------------------------------------------
    # Distribution shape
    # --------------------------------------------------------

    print("\n[ DISTRIBUTION SHAPE ]")
    print(f"Skewness            : {skewness:.4f}")
    print(f"Kurtosis            : {kurtosis:.4f}")

    if skewness > 1:
        skew_interpretation = "Strongly right-skewed"
    elif skewness > 0.5:
        skew_interpretation = "Moderately right-skewed"
    elif skewness < -1:
        skew_interpretation = "Strongly left-skewed"
    elif skewness < -0.5:
        skew_interpretation = "Moderately left-skewed"
    else:
        skew_interpretation = "Approximately symmetric"

    print(f"Skewness interpretation: {skew_interpretation}")

    # --------------------------------------------------------
    # Middle 50% concentration
    # --------------------------------------------------------

    middle_50 = s[(s >= q1) & (s <= q3)]

    print("\n[ MIDDLE 50% CONCENTRATION ]")
    print(f"Q1                  : {q1:.4f}")
    print(f"Q3                  : {q3:.4f}")
    print(f"Values between Q1-Q3: {len(middle_50):,}")
    print(f"Percentage          : {len(middle_50) / n * 100:.2f}%")

    # --------------------------------------------------------
    # Exact value frequency table
    # --------------------------------------------------------

    print("\n[ VALUE FREQUENCY ]")

    value_counts = s.value_counts().sort_index()

    if unique <= 100:
        frequency_table = pd.DataFrame({
            "Value": value_counts.index,
            "Count": value_counts.values,
            "Percentage": (value_counts.values / n * 100)
        })

        frequency_table["Percentage"] = frequency_table["Percentage"].round(2)

        print(frequency_table.to_string(index=False))

    else:
        top_values = (
            s.value_counts()
            .head(20)
            .rename_axis("Value")
            .reset_index(name="Count")
        )

        top_values["Percentage"] = (
            top_values["Count"] / n * 100
        ).round(2)

        print("Top 20 most frequent values:")
        print(top_values.to_string(index=False))

    # ========================================================
    # VISUALIZATION
    # ========================================================

    fig, axes = plt.subplots(
        2,
        2,
        figsize=(17, 12)
    )

    fig.suptitle(
        f"Numerical EDA — {feature}",
        fontsize=20,
        fontweight="bold",
        y=1.02
    )

    # --------------------------------------------------------
    # Plot 1 — Histogram
    # --------------------------------------------------------

    ax = axes[0, 0]

    ax.hist(
        s,
        bins=30,
        color=color,
        edgecolor="white",
        linewidth=1.2,
        alpha=0.85
    )

    ax.axvline(
        mean,
        color=STAT_COLORS["mean"],
        linestyle="--",
        linewidth=2.5,
        label=f"Mean = {mean:.2f}"
    )

    ax.axvline(
        median,
        color=STAT_COLORS["median"],
        linestyle="-.",
        linewidth=2.5,
        label=f"Median = {median:.2f}"
    )

    ax.axvline(
        q1,
        color=STAT_COLORS["q1"],
        linestyle=":",
        linewidth=2.5,
        label=f"Q1 = {q1:.2f}"
    )

    ax.axvline(
        q3,
        color=STAT_COLORS["q3"],
        linestyle=":",
        linewidth=2.5,
        label=f"Q3 = {q3:.2f}"
    )

    ax.set_title(
        "Distribution of Values",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel(feature)
    ax.set_ylabel("Frequency")

    ax.legend(
        fontsize=9,
        frameon=True
    )

    ax.grid(
        alpha=0.25,
        linestyle="--"
    )

    # --------------------------------------------------------
    # Plot 2 — Boxplot
    # --------------------------------------------------------

    ax = axes[0, 1]

    box = ax.boxplot(
        s,
        vert=True,
        patch_artist=True,
        widths=0.45,
        flierprops={
            "marker": "o",
            "markerfacecolor": STAT_COLORS["outlier"],
            "markeredgecolor": STAT_COLORS["outlier"],
            "markersize": 6,
            "alpha": 0.75
        },
        medianprops={
            "color": STAT_COLORS["median"],
            "linewidth": 3
        },
        whiskerprops={
            "color": color,
            "linewidth": 2
        },
        capprops={
            "color": color,
            "linewidth": 2
        }
    )

    box["boxes"][0].set_facecolor(light_color)
    box["boxes"][0].set_edgecolor(color)
    box["boxes"][0].set_linewidth(2)

    ax.set_title(
        "Boxplot & Outliers",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_ylabel(feature)

    ax.grid(
        axis="y",
        alpha=0.25,
        linestyle="--"
    )

    # --------------------------------------------------------
    # Plot 3 — ECDF
    # --------------------------------------------------------

    ax = axes[1, 0]

    sorted_values = np.sort(s.values)
    cumulative = np.arange(1, len(sorted_values) + 1) / len(sorted_values)

    ax.plot(
        sorted_values,
        cumulative,
        color=color,
        linewidth=3
    )

    ax.fill_between(
        sorted_values,
        cumulative,
        alpha=0.18,
        color=color
    )

    ax.axvline(
        median,
        color=STAT_COLORS["median"],
        linestyle="--",
        linewidth=2,
        label=f"Median = {median:.2f}"
    )

    ax.axhline(
        0.50,
        color=STAT_COLORS["mean"],
        linestyle=":",
        linewidth=2,
        label="50%"
    )

    ax.set_title(
        "Cumulative Distribution (ECDF)",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel(feature)
    ax.set_ylabel("Cumulative proportion")

    ax.set_ylim(0, 1.02)

    ax.legend(fontsize=9)

    ax.grid(
        alpha=0.25,
        linestyle="--"
    )

    # --------------------------------------------------------
    # Plot 4 — Range distribution
    # --------------------------------------------------------

    ax = axes[1, 1]

    if minimum != maximum:

        bins = np.linspace(
            minimum,
            maximum,
            11
        )

        counts, edges = np.histogram(
            s,
            bins=bins
        )

        centers = (edges[:-1] + edges[1:]) / 2

        # Different color for every range bin
        bin_colors = [
            FEATURE_COLORS[j % len(FEATURE_COLORS)]
            for j in range(len(counts))
        ]

        bars = ax.bar(
            centers,
            counts,
            width=np.diff(edges) * 0.85,
            color=bin_colors,
            edgecolor="white",
            linewidth=1.2,
            alpha=0.90
        )

        # Add count labels
        for bar, count in zip(bars, counts):

            if count > 0:
                ax.text(
                    bar.get_x() + bar.get_width() / 2,
                    bar.get_height(),
                    f"{count:,}",
                    ha="center",
                    va="bottom",
                    fontsize=9,
                    fontweight="bold"
                )

        ax.set_title(
            "Value Distribution Across Range",
            fontsize=14,
            fontweight="bold"
        )

        ax.set_xlabel(feature)
        ax.set_ylabel("Frequency")

        ax.grid(
            axis="y",
            alpha=0.25,
            linestyle="--"
        )

    else:

        ax.bar(
            [minimum],
            [n],
            color=color,
            edgecolor="white"
        )

        ax.set_title(
            "Constant Numerical Feature",
            fontsize=14,
            fontweight="bold"
        )

        ax.set_xlabel(feature)
        ax.set_ylabel("Frequency")

    # --------------------------------------------------------
    # Final formatting
    # --------------------------------------------------------

    plt.tight_layout()
    plt.show()

    # --------------------------------------------------------
    # Quick interpretation
    # --------------------------------------------------------

    print("\n[ QUICK INTERPRETATION ]")

    print(
        f"• Values range from {minimum:.4f} to {maximum:.4f}."
    )

    print(
        f"• The middle 50% of observations lie between "
        f"{q1:.4f} and {q3:.4f}."
    )

    print(
        f"• The most frequent value is {mode:.4f}, "
        f"appearing {mode_count:,} times ({mode_percentage:.2f}%)."
    )

    if len(outliers) > 0:
        print(
            f"• {len(outliers):,} observations "
            f"({len(outliers) / n * 100:.2f}%) "
            f"are flagged as IQR-based outliers."
        )
    else:
        print("• No IQR-based outliers were detected.")

    print(
        f"• Distribution shape: {skew_interpretation.lower()}."
    )

    print("-" * 100)
```

## All Categorical Features Detail Analysis

```python
# ============================================================
# CATEGORICAL
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Identify categorical features
# ------------------------------------------------------------

categorical_features = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

# Do not analyze target here
if TARGET in categorical_features:
    categorical_features.remove(TARGET)

print("=" * 100)
print("CATEGORICAL EDA")
print("=" * 100)
print(f"Number of categorical features: {len(categorical_features)}")
print(f"Categorical features: {categorical_features}")


# ------------------------------------------------------------
# 2. Color palettes
# ------------------------------------------------------------

CATEGORY_PALETTE = [
    "#4C78A8",  # blue
    "#F58518",  # orange
    "#54A24B",  # green
    "#E45756",  # red
    "#B279A2",  # purple
    "#72B7B2",  # teal
    "#ECA82C",  # gold
    "#FF9DA6",  # pink
    "#9D755D",  # brown
    "#79706E",  # gray
    "#59A14F",  # green
    "#B07AA1",  # violet
]

RARE_COLOR = "#D81B60"
MISSING_COLOR = "#7E57C2"
DOMINANT_COLOR = "#E63946"


# ------------------------------------------------------------
# 3. Analyze every categorical feature
# ------------------------------------------------------------

for feature_index, feature in enumerate(categorical_features):

    s = df[feature]

    # --------------------------------------------------------
    # Basic information
    # --------------------------------------------------------

    total_rows = len(s)
    missing_count = int(s.isna().sum())
    non_missing_count = int(s.notna().sum())

    unique_count = s.nunique(dropna=True)

    # Treat missing values separately for frequency analysis
    analysis_s = s.astype("object").where(
        s.notna(),
        "MISSING"
    )

    value_counts = analysis_s.value_counts()

    percentages = (
        value_counts / total_rows * 100
    )

    # --------------------------------------------------------
    # Dominant category
    # --------------------------------------------------------

    dominant_category = value_counts.index[0]
    dominant_count = int(value_counts.iloc[0])
    dominant_percentage = float(percentages.iloc[0])

    # --------------------------------------------------------
    # Rare categories
    #
    # Rare = category occurring in <= 1% of dataset
    # --------------------------------------------------------

    rare_threshold = total_rows * 0.01

    rare_categories = value_counts[
        value_counts <= rare_threshold
    ]

    # --------------------------------------------------------
    # Header
    # --------------------------------------------------------

    print("\n\n")
    print("=" * 110)
    print(
        f"CATEGORICAL FEATURE {feature_index + 1}: {feature}"
    )
    print("=" * 110)

    # --------------------------------------------------------
    # Basic information
    # --------------------------------------------------------

    print("\n[ BASIC INFORMATION ]")

    print(f"Total observations      : {total_rows:,}")
    print(f"Non-missing values      : {non_missing_count:,}")
    print(f"Missing values          : {missing_count:,}")
    print(
        f"Missing percentage      : "
        f"{missing_count / total_rows * 100:.2f}%"
    )
    print(f"Unique categories       : {unique_count:,}")

    # --------------------------------------------------------
    # Dominant category
    # --------------------------------------------------------

    print("\n[ DOMINANT CATEGORY ]")

    print(f"Most frequent category : {dominant_category}")
    print(f"Frequency              : {dominant_count:,}")
    print(f"Percentage             : {dominant_percentage:.2f}%")

    # --------------------------------------------------------
    # Category frequency table
    # --------------------------------------------------------

    print("\n[ COMPLETE CATEGORY FREQUENCY TABLE ]")

    frequency_table = pd.DataFrame({
        "Category": value_counts.index.astype(str),
        "Count": value_counts.values,
        "Percentage": percentages.values.round(2)
    })

    print(
        frequency_table.to_string(index=False)
    )

    # --------------------------------------------------------
    # Rare categories
    # --------------------------------------------------------

    print("\n[ RARE CATEGORY ANALYSIS ]")

    if len(rare_categories) > 0:

        rare_percentage = (
            rare_categories.sum()
            / total_rows
            * 100
        )

        print(
            f"Rare-category threshold : <= 1% "
            f"({rare_threshold:.0f} observations)"
        )

        print(
            f"Number of rare categories: "
            f"{len(rare_categories)}"
        )

        print(
            f"Observations in rare categories: "
            f"{rare_categories.sum():,} "
            f"({rare_percentage:.2f}%)"
        )

        rare_table = pd.DataFrame({
            "Category": rare_categories.index.astype(str),
            "Count": rare_categories.values,
            "Percentage": (
                rare_categories.values
                / total_rows
                * 100
            ).round(2)
        })

        print("\nRare categories:")
        print(
            rare_table.to_string(index=False)
        )

    else:

        print("No categories occur at or below 1%.")


    # --------------------------------------------------------
    # Missing value analysis
    # --------------------------------------------------------

    print("\n[ MISSING VALUE ANALYSIS ]")

    if missing_count > 0:

        print(
            f"Missing values found: {missing_count:,} "
            f"({missing_count / total_rows * 100:.2f}%)"
        )

    else:

        print("No missing values detected.")


    # ========================================================
    # VISUALIZATION
    # ========================================================

    # Use a different main color for every feature
    feature_color = CATEGORY_PALETTE[
        feature_index % len(CATEGORY_PALETTE)
    ]

    # Number of categories
    n_categories = len(value_counts)

    # --------------------------------------------------------
    # For many categories, show top 20 visually
    # but preserve COMPLETE table above.
    # --------------------------------------------------------

    if n_categories > 20:

        plot_counts = value_counts.head(20)

        plot_title_suffix = " — Top 20 Categories"

    else:

        plot_counts = value_counts

        plot_title_suffix = ""


    # --------------------------------------------------------
    # Create figure
    # --------------------------------------------------------

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(18, 7)
    )

    fig.suptitle(
        f"Categorical EDA — {feature}",
        fontsize=20,
        fontweight="bold",
        y=1.02
    )


    # ========================================================
    # Plot 1 — Colorful horizontal bar chart
    # ========================================================

    ax = axes[0]

    categories = plot_counts.index.astype(str)[::-1]
    counts = plot_counts.values[::-1]

    # Give every category a different color
    bar_colors = [
        CATEGORY_PALETTE[
            j % len(CATEGORY_PALETTE)
        ]
        for j in range(len(categories))
    ]

    bars = ax.barh(
        categories,
        counts,
        color=bar_colors,
        edgecolor="white",
        linewidth=1.2,
        alpha=0.90
    )

    # Add count labels
    for bar, count in zip(bars, counts):

        ax.text(
            bar.get_width(),
            bar.get_y() + bar.get_height() / 2,
            f" {count:,}",
            va="center",
            fontsize=9,
            fontweight="bold"
        )

    ax.set_title(
        f"Category Frequency{plot_title_suffix}",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel("Number of observations")
    ax.set_ylabel(feature)

    ax.grid(
        axis="x",
        alpha=0.25,
        linestyle="--"
    )


    # ========================================================
    # Plot 2 — Percentage distribution
    # ========================================================

    ax = axes[1]

    percentage_values = (
        plot_counts / total_rows * 100
    )

    categories = plot_counts.index.astype(str)

    percentage_colors = [
        CATEGORY_PALETTE[
            j % len(CATEGORY_PALETTE)
        ]
        for j in range(len(categories))
    ]

    bars = ax.bar(
        categories,
        percentage_values,
        color=percentage_colors,
        edgecolor="white",
        linewidth=1.2,
        alpha=0.90
    )

    # Add percentage labels
    for bar, percentage in zip(
        bars,
        percentage_values
    ):

        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f"{percentage:.1f}%",
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold"
        )

    ax.set_title(
        f"Category Percentage Distribution{plot_title_suffix}",
        fontsize=14,
        fontweight="bold"
    )

    ax.set_xlabel(feature)
    ax.set_ylabel("Percentage (%)")

    ax.tick_params(
        axis="x",
        rotation=45
    )

    ax.grid(
        axis="y",
        alpha=0.25,
        linestyle="--"
    )


    # --------------------------------------------------------
    # Final formatting
    # --------------------------------------------------------

    plt.tight_layout()
    plt.show()


    # ========================================================
    # Additional visualization for small category sets
    # ========================================================

    if 2 <= n_categories <= 8:

        fig, ax = plt.subplots(
            figsize=(9, 8)
        )

        pie_colors = [
            CATEGORY_PALETTE[
                j % len(CATEGORY_PALETTE)
            ]
            for j in range(n_categories)
        ]

        wedges, texts, autotexts = ax.pie(
            value_counts.values,
            labels=[
                str(x)
                for x in value_counts.index
            ],
            autopct="%1.1f%%",
            startangle=90,
            colors=pie_colors,
            wedgeprops={
                "edgecolor": "white",
                "linewidth": 2
            }
        )

        for autotext in autotexts:
            autotext.set_fontweight("bold")
            autotext.set_fontsize(10)

        ax.set_title(
            f"{feature} — Proportion of Categories",
            fontsize=15,
            fontweight="bold"
        )

        plt.tight_layout()
        plt.show()


    # ========================================================
    # Quick interpretation
    # ========================================================

    print("\n[ QUICK INTERPRETATION ]")

    print(
        f"• {feature} contains "
        f"{unique_count:,} unique categories."
    )

    print(
        f"• The most frequent category is "
        f"'{dominant_category}', appearing "
        f"{dominant_count:,} times "
        f"({dominant_percentage:.2f}%)."
    )

    if dominant_percentage >= 70:

        print(
            "• The distribution is highly concentrated "
            "in the dominant category."
        )

    elif dominant_percentage >= 50:

        print(
            "• The distribution is moderately concentrated "
            "in the dominant category."
        )

    else:

        print(
            "• No single category accounts for more than "
            "half of the observations."
        )

    if len(rare_categories) > 0:

        print(
            f"• {len(rare_categories)} categories occur "
            f"in <= 1% of observations."
        )

    else:

        print(
            "• No category occurs in <= 1% "
            "of observations."
        )

    if missing_count > 0:

        print(
            f"• Missing values represent "
            f"{missing_count / total_rows * 100:.2f}% "
            f"of the dataset."
        )

    else:

        print("• No missing categorical values detected.")

    print("-" * 110)
```

## Target Variable Analysis

## CATEGORY-WISE STRESS LEVEL

```python
# ============================================================
# CATEGORY-WISE STRESS LEVEL ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

TARGET = "Stress_Detection"

categorical_features = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

if TARGET in categorical_features:
    categorical_features.remove(TARGET)

print("Categorical features:")
print(categorical_features)
print()


for feature in categorical_features:

    print("\n" + "=" * 90)
    print(f"FEATURE: {feature}")
    print("=" * 90)

    # --------------------------------------------------------
    # 1. Count of each stress level within every category
    # --------------------------------------------------------
    count_table = pd.crosstab(
        df[feature],
        df[TARGET],
        margins=False
    )

    print("\nSTRESS COUNT BY CATEGORY:")
    display(count_table)

    # --------------------------------------------------------
    # 2. Percentage of stress levels within each category
    # --------------------------------------------------------
    percentage_table = pd.crosstab(
        df[feature],
        df[TARGET],
        normalize="index"
    ) * 100

    percentage_table = percentage_table.round(2)

    print("\nSTRESS PERCENTAGE WITHIN EACH CATEGORY:")
    display(percentage_table)

    # --------------------------------------------------------
    # 3. Identify highest stress category
    # --------------------------------------------------------
    if "High" in percentage_table.columns:
        highest_high_category = percentage_table["High"].idxmax()
        highest_high_percentage = percentage_table["High"].max()

        print(
            f"\nHighest HIGH-stress category: "
            f"{highest_high_category} "
            f"({highest_high_percentage:.2f}%)"
        )

    # --------------------------------------------------------
    # 4. Identify lowest stress category
    # --------------------------------------------------------
    if "Low" in percentage_table.columns:
        highest_low_category = percentage_table["Low"].idxmax()
        highest_low_percentage = percentage_table["Low"].max()

        print(
            f"Highest LOW-stress category: "
            f"{highest_low_category} "
            f"({highest_low_percentage:.2f}%)"
        )

    # --------------------------------------------------------
    # 5. Identify highest Medium-stress category
    # --------------------------------------------------------
    if "Medium" in percentage_table.columns:
        highest_medium_category = percentage_table["Medium"].idxmax()
        highest_medium_percentage = percentage_table["Medium"].max()

        print(
            f"Highest MEDIUM-stress category: "
            f"{highest_medium_category} "
            f"({highest_medium_percentage:.2f}%)"
        )

    # --------------------------------------------------------
    # 6. Visualization
    # --------------------------------------------------------
    percentage_table.plot(
        kind="bar",
        figsize=(11, 6),
        width=0.8
    )

    plt.title(
        f"Stress Level Distribution by {feature}",
        fontsize=14,
        fontweight="bold"
    )

    plt.xlabel(feature)
    plt.ylabel("Percentage (%)")
    plt.xticks(rotation=45, ha="right")
    plt.legend(title="Stress Level")
    plt.grid(axis="y", alpha=0.25)
    plt.tight_layout()
    plt.show()
```

## ALL NUMERICAL FEATURES × STRESS LEVEL

```python
# ============================================================
# ALL NUMERICAL FEATURES × STRESS LEVEL
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgb

TARGET = "Stress_Detection"

# ------------------------------------------------------------
# Numerical features
# ------------------------------------------------------------
numeric_features = df.select_dtypes(include=np.number).columns.tolist()

if TARGET in numeric_features:
    numeric_features.remove(TARGET)

print(f"Total numerical features: {len(numeric_features)}")
print(numeric_features)


# ------------------------------------------------------------
# Different base colors for different numerical features
# ------------------------------------------------------------
FEATURE_COLORS = [
    "#4C78A8",   # Blue
    "#59A14F",   # Green
    "#9C6ADE",   # Purple
    "#F28E2B",   # Orange
    "#E15759",   # Red
    "#76B7B2",   # Teal
    "#EDC948",   # Yellow
    "#B07AA1",   # Violet
    "#FF9DA7",   # Pink
    "#8C6D31",   # Brown
    "#17BECF",   # Cyan
    "#9467BD"    # Deep purple
]


# ------------------------------------------------------------
# Create lighter / darker shades of a base color
# ------------------------------------------------------------
def make_shades(hex_color):

    rgb = np.array(to_rgb(hex_color))

    # Low = light
    low = rgb + (1 - rgb) * 0.60

    # Medium = normal
    medium = rgb

    # High = dark
    high = rgb * 0.65

    return [
        tuple(low),
        tuple(medium),
        tuple(high)
    ]


STRESS_ORDER = ["Low", "Medium", "High"]


# ============================================================
# ANALYSIS FOR EVERY NUMERICAL FEATURE
# ============================================================

for feature_index, feature in enumerate(numeric_features):

    print("\n" + "=" * 100)
    print(f"NUMERICAL FEATURE: {feature}")
    print("=" * 100)

    data = df[[feature, TARGET]].dropna().copy()

    # --------------------------------------------------------
    # Basic range
    # --------------------------------------------------------
    print(f"Minimum : {data[feature].min():.2f}")
    print(f"Maximum : {data[feature].max():.2f}")
    print(f"Range   : {data[feature].max() - data[feature].min():.2f}")

    # --------------------------------------------------------
    # Statistics by stress level
    # --------------------------------------------------------
    stats = data.groupby(TARGET)[feature].agg(
        Count="count",
        Minimum="min",
        Q1=lambda x: x.quantile(0.25),
        Median="median",
        Mean="mean",
        Q3=lambda x: x.quantile(0.75),
        Maximum="max",
        Std="std"
    )

    stats = stats.reindex(
        [x for x in STRESS_ORDER if x in stats.index]
    ).round(2)

    print("\nSTATISTICS BY STRESS LEVEL:")
    display(stats)

    # --------------------------------------------------------
    # Create 5 numerical ranges
    # --------------------------------------------------------
    min_value = data[feature].min()
    max_value = data[feature].max()

    bins = np.linspace(
        min_value,
        max_value,
        6
    )

    bins = np.unique(bins)

    if len(bins) < 3:
        print("Not enough unique values to create ranges.")
        continue

    data["Numerical_Range"] = pd.cut(
        data[feature],
        bins=bins,
        include_lowest=True
    )

    # --------------------------------------------------------
    # Count table
    # --------------------------------------------------------
    count_table = pd.crosstab(
        data["Numerical_Range"],
        data[TARGET]
    )

    count_table = count_table.reindex(
        columns=STRESS_ORDER,
        fill_value=0
    )

    print("\nCOUNT BY NUMERICAL RANGE:")
    display(count_table)

    # --------------------------------------------------------
    # Percentage of entire dataset
    # --------------------------------------------------------
    overall_percentage = (
        count_table / len(data)
    ) * 100

    overall_percentage = overall_percentage.round(2)

    print("\nPERCENTAGE OF ENTIRE DATASET:")
    display(overall_percentage)

    # --------------------------------------------------------
    # Dominant stress level
    # --------------------------------------------------------
    print("\nDOMINANT STRESS LEVEL BY RANGE:")

    for numerical_range in count_table.index:

        row = count_table.loc[numerical_range]

        dominant_level = row.idxmax()
        dominant_count = row.max()

        print(
            f"{str(numerical_range):30s} → "
            f"{dominant_level} "
            f"({dominant_count} people)"
        )

    # ========================================================
    # DIFFERENT COLOR FOR EACH NUMERICAL FEATURE
    # ========================================================

    base_color = FEATURE_COLORS[
        feature_index % len(FEATURE_COLORS)
    ]

    shades = make_shades(base_color)

    colors = {
        "Low": shades[0],
        "Medium": shades[1],
        "High": shades[2]
    }

    # --------------------------------------------------------
    # BAR CHART
    # --------------------------------------------------------
    ax = count_table.plot(
        kind="bar",
        figsize=(12, 6),
        color=[
            colors["Low"],
            colors["Medium"],
            colors["High"]
        ],
        edgecolor="white",
        linewidth=0.8,
        width=0.8
    )

    plt.title(
        f"{feature}: Stress Level by Numerical Range",
        fontsize=15,
        fontweight="bold"
    )

    plt.xlabel(
        f"{feature} Range",
        fontsize=11
    )

    plt.ylabel(
        "Number of People",
        fontsize=11
    )

    plt.xticks(
        rotation=45,
        ha="right"
    )

    plt.legend(
        title="Stress Level"
    )

    plt.grid(
        axis="y",
        alpha=0.20
    )

    plt.tight_layout()
    plt.show()


print("\n" + "=" * 100)
print("ALL NUMERICAL FEATURES ANALYSIS COMPLETED")
print("=" * 100)
```
