### Import Libraries


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 100)

print("Libraries imported successfully.")
```

### Load and verify the raw dataset

```python
# Load and verify the raw dataset

PROJECT_ROOT = Path.cwd().parent
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw" / "extended_stress_detection_data.csv"

df = pd.read_csv(RAW_DATA_PATH)

print("Dataset loaded successfully.")
print(f"File   : {RAW_DATA_PATH.name}")
print(f"Shape  : {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Exists : {RAW_DATA_PATH.exists()}")
```

```python
# Initial inspection of the raw dataset

display(df.head(10))

print("\nColumn names:")
for i, column in enumerate(df.columns, 1):
    print(f"{i:2}. {column}")

print(f"\nDataset shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# Data quality audit

audit = pd.DataFrame({
    "Data Type": df.dtypes.astype(str),
    "Non-Null": df.notna().sum(),
    "Missing": df.isna().sum(),
    "Unique Values": df.nunique(dropna=True)
})

print(f"Total records: {len(df):,}")
print(f"Total features: {df.shape[1]}")
print(f"Duplicate rows: {df.duplicated().sum():,}")

display(audit)
```

### Feature and Target Classification

```python
# Classify features and target

TARGET = "Stress_Detection"

numeric_features = df.select_dtypes(include=np.number).columns.tolist()
categorical_features = df.select_dtypes(exclude=np.number).columns.tolist()
categorical_features.remove(TARGET)

print(f"Target variable: {TARGET}")
print(f"Input features: {df.shape[1] - 1}")
print(f"Numerical features: {len(numeric_features)}")
print(f"Categorical features: {len(categorical_features)}")

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
# Target class distribution

target_summary = (
    df[TARGET]
    .value_counts()
    .rename_axis("Stress Level")
    .to_frame("Count")
)

target_summary["Percentage"] = (
    target_summary["Count"] / len(df) * 100
).round(2)

display(target_summary)

print(f"Number of classes: {df[TARGET].nunique()}")
print(f"Classes: {sorted(df[TARGET].unique())}")
```

```python
# Numerical and categorical feature summary

print("NUMERICAL FEATURES")
display(df[numeric_features].describe().T.round(2))

print("\nCATEGORICAL FEATURES")
categorical_summary = pd.DataFrame({
    "Unique Values": df[categorical_features].nunique(),
    "Most Frequent": df[categorical_features].mode().iloc[0],
    "Frequency": df[categorical_features].apply(lambda col: col.value_counts().iloc[0])
})

display(categorical_summary)
```

## Raw Dataset Audit — Summary

The raw Human Stress Detection Extended Dataset contains **3,000 records and 22 columns**, consisting of **21 input features and one target variable (`Stress_Detection`)**.

The input features include **13 numerical variables** and **8 categorical variables** covering demographic, lifestyle, sleep, occupational, behavioral, and physiological characteristics.

The target variable, `Stress_Detection`, contains three classes:

- **Medium:** 1,258 records (41.93%)
- **High:** 1,118 records (37.27%)
- **Low:** 624 records (20.80%)

The dataset contains **no missing values** and **no exact duplicate rows**.

The categorical variables include `Gender`, `Occupation`, `Marital_Status`, `Wake_Up_Time`, `Bed_Time`, `Smoking_Habit`, `Meditation_Practice`, and `Exercise_Type`. `Occupation` has a relatively high cardinality with 169 unique categories.

Several numerical variables contain values outside their expected intuitive ranges, including negative values in `Physical_Activity`, `Screen_Time`, `Travel_Time`, and `Social_Interactions`. These values are retained in the raw dataset and will be investigated during the exploratory data analysis stage.

The target distribution shows moderate class imbalance, with the `Low` stress class representing the smallest group.

### Audit Conclusion

The raw dataset is structurally complete, with no missing or duplicate records. However, class distribution, unusual numerical values, and high-cardinality categorical variables require further investigation during EDA before preprocessing and graph construction.
